```python
# Simple Blood Bank Management System

blood_bank = {
    "A+": 5,
    "A-": 3,
    "B+": 4,
    "B-": 2,
    "AB+": 3,
    "AB-": 1,
    "O+": 6,
    "O-": 2
}

while True:
    print("\n===== BLOOD BANK MANAGEMENT =====")
    print("1. View Blood Stock")
    print("2. Search Blood")
    print("3. Add Blood")
    print("4. Request Blood")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nBlood Stock:")
        for group, units in blood_bank.items():
            print(group, ":", units, "units")

    elif choice == "2":
        group = input("Enter blood group: ").upper()

        if group in blood_bank:
            print(group, "available:", blood_bank[group], "units")
        else:
            print("Invalid blood group")

    elif choice == "3":
        group = input("Enter blood group: ").upper()
        units = int(input("Enter number of units: "))

        if group in blood_bank:
            blood_bank[group] += units
            print("Blood added successfully!")
        else:
            print("Invalid blood group")

    elif choice == "4":
        group = input("Enter blood group: ").upper()
        units = int(input("Enter required units: "))

        if group in blood_bank:
            if blood_bank[group] >= units:
                blood_bank[group] -= units
                print("Blood request approved!")
            else:
                print("Not enough blood available")
        else:
            print("Invalid blood group")

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid choice")
```
